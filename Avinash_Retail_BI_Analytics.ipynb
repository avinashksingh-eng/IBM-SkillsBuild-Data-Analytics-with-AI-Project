```python
import pandas as pd
import numpy as np

# Mock Data Generation
np.random.seed(42)
n = 100
df = pd.DataFrame({
    'OrderID': range(1001, 1001 + n),
    'Date': pd.date_range(start='2024-01-01', periods=n, freq='D'),
    'Category': np.random.choice(['Electronics', 'Clothing', 'Home', 'Books'], n),
    'Region': np.random.choice(['North', 'South', 'East', 'West'], n),
    'SalesAmount': np.random.randint(500, 5000, n),
    'Quantity': np.random.randint(1, 5, n),
    'CustomerStatus': np.random.choice(['Active', 'Active', 'Churn Risk'], n)
})

# Analysis
print("=== KEY PERFORMANCE INDICATORS ===")
print("Total Revenue:", df['SalesAmount'].sum())
print("Total Orders:", df['OrderID'].nunique())
print("Churn Risk Customers:", (df['CustomerStatus'] == 'Churn Risk').sum())

print("\n=== REVENUE BY CATEGORY ===")
print(df.groupby('Category')['SalesAmount'].sum())
```
